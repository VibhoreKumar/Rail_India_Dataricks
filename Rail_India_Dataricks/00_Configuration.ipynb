Delta lake storage configuration

```python
base_path='/Volumes/workspace/default'

#raw csv input
raw_path=f'{base_path}/railway_raw'

#delta lake storage locations
bronze_path=f'{base_path}/railway_bronze'
silver_path=f'{base_path}/railway_silver'
gold_path=f'{base_path}/railway_gold'

#configured path
print('raw csv :',raw_path)
print("Bronze Delta Path:", bronze_path)
print("Silver Delta Path:", silver_path)
print("Gold Delta Path:", gold_path)
```
